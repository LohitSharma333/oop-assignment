### What is Object-Oriented Programming (OOP)?

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of 'objects', which contain data and methods. It promotes reusability, modularity, and abstraction.

### What is a class in OOP?

A class in OOP is a blueprint for creating objects. It defines attributes and methods common to all instances of the class.

### What is an object in OOP?

An object is an instance of a class. It represents a real-world entity with state and behavior.

### What is the difference between abstraction and encapsulation?

Abstraction hides complexity by showing only the relevant details, while encapsulation restricts direct access to some of an object's components to protect the integrity of the data.

### What are dunder methods in Python?

Dunder methods (short for 'double underscore') are special methods in Python like __init__, __str__, and __repr__ that allow customization of class behavior.

### Explain the concept of inheritance in OOP.

Inheritance allows a class (child) to inherit properties and behaviors from another class (parent), promoting code reuse.

### What is polymorphism in OOP?

Polymorphism allows methods to have the same name but behave differently depending on the object calling them.

### How is encapsulation achieved in Python?

Encapsulation is achieved using private variables (by prefixing with an underscore) and providing public methods for access and modification.

### What is a constructor in Python?

A constructor in Python is the __init__ method. It is automatically called when an object is created.

### What are class and static methods in Python?

@classmethod is bound to the class and not the instance; @staticmethod does not access the class or instance.

### What is method overloading in Python?

Method overloading refers to having multiple methods with the same name but different parameters (not natively supported in Python like in Java).

### What is method overriding in OOP?

Method overriding occurs when a subclass provides a specific implementation of a method already defined in its parent class.

### What is a property decorator in Python?

The @property decorator allows a method to be accessed like an attribute, commonly used for getter methods.

### Why is polymorphism important in OOP?

Polymorphism allows for flexible and reusable code, especially in scenarios where objects can take on many forms.

### What is an abstract class in Python?

An abstract class contains one or more abstract methods and cannot be instantiated. It serves as a template for other classes.

### What are the advantages of OOP?

Advantages of OOP include modularity, reusability, scalability, and ease of maintenance.

### What is the difference between a class variable and an instance variable?

A class variable is shared among all instances of a class, while an instance variable is unique to each instance.

### What is multiple inheritance in Python?

Multiple inheritance allows a class to inherit from more than one base class.

### Explain the purpose of ‘__str__’ and ‘__repr__’ methods in Python.

__str__ returns a user-friendly string representation of the object, while __repr__ returns a developer-friendly string.

### What is the significance of the ‘super()’ function in Python?

The super() function allows access to methods from a parent or sibling class.

### What is the significance of the __del__ method in Python?

__del__ is a destructor method that is called when an object is about to be destroyed.

### What is the difference between @staticmethod and @classmethod in Python?

@staticmethod does not access class or instance data; @classmethod receives the class as the first argument.

### How does polymorphism work in Python with inheritance?

In Python, polymorphism through inheritance allows base class references to call overridden methods in derived classes.

### What is method chaining in Python OOP?

Method chaining is returning self from methods so multiple calls can be made in a single line.

### What is the purpose of the __call__ method in Python?

__call__ allows an instance of a class to be called as a function.

### 1. Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog that overrides the speak() method to print 'Bark!'.

```
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Bark!")

Dog().speak()
```

### 2. Create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method.

```
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
```

### 3. Implement multi-level inheritance with Vehicle -> Car -> ElectricCar, adding a battery attribute.

```
class Vehicle:
    def __init__(self, v_type):
        self.type = v_type

class Car(Vehicle):
    def __init__(self, v_type, brand):
        super().__init__(v_type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, v_type, brand, battery):
        super().__init__(v_type, brand)
        self.battery = battery
```

### 4. Demonstrate polymorphism using a base class Bird and two derived classes Sparrow and Penguin with overridden fly() methods.

```
class Bird:
    def fly(self):
        print("Bird is flying")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high")

class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly")
```

### 5. Demonstrate encapsulation with a class BankAccount having private attributes and methods to deposit, withdraw, and check balance.

```
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount

    def check_balance(self):
        return self.__balance
```

### 6. Demonstrate runtime polymorphism with a play() method in Instrument and its overrides in Guitar and Piano.

```
class Instrument:
    def play(self):
        print("Playing instrument")

class Guitar(Instrument):
    def play(self):
        print("Playing guitar")

class Piano(Instrument):
    def play(self):
        print("Playing piano")
```

### 7. Create a class MathOperations with a class method add_numbers() and a static method subtract_numbers().

```
class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b
```

### 8. Implement a class Person with a class method to count the total number of persons created.

```
class Person:
    count = 0

    def __init__(self):
        Person.count += 1

    @classmethod
    def total_persons(cls):
        return cls.count
```

### 9. Create a class Fraction with overridden __str__ method to display 'numerator/denominator'.

```
class Fraction:
    def __init__(self, num, denom):
        self.numerator = num
        self.denominator = denom

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
```

### 10. Demonstrate operator overloading in a Vector class to add two vectors.

```
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"
```

### 12. Implement a class Student with name and grades and a method average_grade().

```
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        return sum(self.grades) / len(self.grades)
```

### 13. Create a class Rectangle with set_dimensions() and area() methods.

```
class Rectangle:
    def set_dimensions(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
```

### 14. Create a class Employee with calculate_salary() and a derived class Manager with bonus.

```
class Employee:
    def calculate_salary(self, hours, rate):
        return hours * rate

class Manager(Employee):
    def calculate_salary(self, hours, rate, bonus):
        return super().calculate_salary(hours, rate) + bonus
```

### 15. Create a class Product with total_price() method.

```
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity
```

### 16. Create an abstract class Animal with sound() method; derive Cow and Sheep.

```
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo")

class Sheep(Animal):
    def sound(self):
        print("Baa")
```

### 17. Create a class Book with title, author, and year_published, and get_book_info() method.

```
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year_published = year

    def get_book_info(self):
        return f"{self.title} by {self.author}, published in {self.year_published}"
```

### 18. Create a class House with address and price, and a derived class Mansion with number_of_rooms.

```
class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, rooms):
        super().__init__(address, price)
        self.number_of_rooms = rooms
```
